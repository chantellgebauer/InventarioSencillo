Creamos un pequeño programa que descuente el inventario de una hamburgueseria

```python
# Definimos la variable inventario y le damos un valor inicial de 100 (tipo entero)
inventario = 100

# Mostramos al usuario el mensaje inicial con las existencias totales
print(f"El inventario contiene {inventario} hamburguesas")

# Bucle principal: se ejecutará de forma continua mientras queden hamburguesas (> 0)
while inventario > 0:

    # Alerta visual si las existencias bajan de 10 unidades
    if inventario <= 10:
        print("El inventario esta casi vacio: CUIDADO 📌")

    # Bloque try-except para evitar que el programa falle si no introducen un número
    try:
        # Capturamos la cantidad solicitada y la convertimos a entero
        num_hamburguesas = int(input("¿Cuantas hamburguesas quiere el cliente? "))
    except ValueError:
        # Si introduce letras o símbolos, atrapamos el error y avisamos al usuario
        print("Por favor, introduce un número válido.")
        continue # Volvemos al inicio del bucle sin ejecutar lo de abajo

    # Validación: el pedido no puede ser menor o igual a cero
    if num_hamburguesas <= 0:
        print("Por favor, introduce un número positivo de hamburguesas.")
        continue # Volvemos al inicio del bucle

    # Validación: el pedido no puede superar las existencias actuales
    if num_hamburguesas > inventario:
        print("No hay suficientes hamburguesas. Solo hay", inventario, "hamburguesas disponibles")
    else:
        # Si todo está correcto, restamos las hamburguesas vendidas del inventario
        inventario -= num_hamburguesas
        # Confirmamos la venta y mostramos el stock restante
        print(f"El cliente ha pedido {num_hamburguesas} hamburguesas. El inventario ahora tiene {inventario} hamburguesas")

# Este mensaje solo se muestra cuando el bucle 'while' termina (el inventario llega a 0)
print("Inventario agotado. Vuelve mañana!")
```
